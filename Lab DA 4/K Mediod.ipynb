```python
import numpy as np
import pandas as pd
from sklearn_extra.cluster import KMedoids
printDN()
```

```python
data = np.array([
[9,6],
[10,4],
[4,4],
[5,8],
[3,8],
[2,5],
[8,5],
[4,6],
[8,4],
[9,3]
])

df = pd.DataFrame(data, columns=["X","Y"])
printDN()
```

```python
initial_medoids = np.array([
[8,4],
[4,6]
])
printDN()
```

```python
kmedoids = KMedoids(
    n_clusters=2,
    init=initial_medoids,
    method='pam'
)

kmedoids.fit(df)
printDN()
```

```python
df["Cluster"] = kmedoids.labels_

print("Clustered Data:")
print(df)
printDN()
```

```python
print("\nFinal Medoids:")
print(kmedoids.cluster_centers_)
printDN()
```
